```python
df['horsepower'] = df.horsepower.fillna(max(df.horsepower))
df.describe()
```

```python
X = df[['vehicle_weight','model_year']].sort_values(by=['model_year'], ascending=True).head(7)
```

```python
XTX = X.T.dot(X)
XTX_inv = np.linalg.inv(XTX)

y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

w = XTX_inv.dot(X.T).dot(y)

print("The sum of the elements in w is:", w.sum())
```
